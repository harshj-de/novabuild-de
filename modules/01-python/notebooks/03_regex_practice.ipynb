```python
import re
```

```python
re.findall(r"log","system log started")
```

```python
re.findall(r'[0-9]','order123value45')
```

```python
re.findall(r"[A-Z]","ErrorIDABC123")
```

```python
re.findall(r"([A-Za-z])","Data_Engineer 2024!")
```

```python
re.findall(r"\d+", "ID45A900")
```

```python
re.findall(r"\d{3,}","ID123 A45 B6789")
```

```python
re.findall(r"(\+\d+)-(\d{10,})","abc+91-9876543210")
```

```python
re.findall(r"\d+(?=kg|cm)", "100kg 50INR, 500cm")
```

```python
re.fullmatch(
    r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}",
    "user@mail.com")
```
