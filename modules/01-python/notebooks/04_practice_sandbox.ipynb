**The Full Code for Creating a Data Set**

```python
import pandas as pd
import random
from datetime import datetime, timedelta

# -----------------------------
# SETUP
# -----------------------------
random.seed(42)

def random_date(start, end):
  return start + timedelta(days=random.randint(0, (end - start).days))

start_date = datetime(2023, 1, 1)
end_date = datetime(2024, 12, 31)

# -----------------------------
# 1. CUSTOMERS
# -----------------------------
first_names = ["Priya", "Arjun", "Rahul"]
last_names = ["Shah", "Patel", "Mehta"]
countries = ["India", "USA", "UK"]

customers = []

for i in range(10):
  name = random.choice(first_names) + " " + random.choice(last_names)
  country = random.choice(countries)
  spent = random.randint(1000, 50000)

    # Data issues
  if i % 3 == 0:
    name = "  " + name + "  "
  if i % 5 == 0:
    spent = None

  customers.append({
      "customer_id": i,
      "name": name,
      "country": country,
      "total_spent": spent
    })

df_customers = pd.DataFrame(customers)
df_customers.to_csv("customers.csv", index=False)

# -----------------------------
# 2. PRODUCTS
# -----------------------------
products = []

categories = ["Electronics", "Clothing", "Food"]

for i in range(5):
    product_id = 100 + i
    category = random.choice(categories)
    price = random.randint(100, 10000)

    products.append({
        "product_id": product_id,
        "category": category,
        "price": price
    })

df_products = pd.DataFrame(products)
df_products.to_csv("products.csv", index=False)

# -----------------------------
# 3. ORDERS
# -----------------------------
orders = []

customer_ids = [c["customer_id"] for c in customers]
product_ids = [p["product_id"] for p in products]

status_list = ["pending", "shipped", "delivered"]

for i in range(10):
    order_id = 1000 + i

    customer_id = random.choice(customer_ids)
    product_id = random.choice(product_ids)

    amount = random.randint(200, 20000)
    status = random.choice(status_list)

    order_date = random_date(start_date, end_date)

    # Delivery logic
    if status in ["shipped", "delivered"]:
        delivery_date = order_date + timedelta(days=random.randint(1, 7))
    else:
        delivery_date = None

    # Data issues
    if i % 4 == 0:
        amount = None

    orders.append({
        "order_id": order_id,
        "customer_id": customer_id,
        "product_id": product_id,
        "amount": amount,
        "status": status,
        "order_date": order_date.strftime("%Y-%m-%d"),
        "delivery_date": delivery_date.strftime("%Y-%m-%d") if delivery_date else None
    })

df_orders = pd.DataFrame(orders)
df_orders.to_csv("orders.csv", index=False)

# -----------------------------
# 4. PAYMENTS
# -----------------------------
payments = []

for order in orders:
    if random.random() > 0.1:  # 90% chance payment exists

        pay_date = datetime.strptime(order["order_date"], "%Y-%m-%d") + timedelta(days=random.randint(0, 2))

        payments.append({
            "payment_id": f"PAY_{order['order_id']}",
            "order_id": order["order_id"],
            "customer_id": order["customer_id"],
            "amount": order["amount"],
            "status": random.choice(["success", "failed"]),
            "payment_date": pay_date.strftime("%Y-%m-%d")
        })

df_payments = pd.DataFrame(payments)
df_payments.to_csv("payments.csv", index=False)

# -----------------------------
# SUMMARY
# -----------------------------
print("Files created:")
print("customers.csv")
print("products.csv")
print("orders.csv")
print("payments.csv")
```
