```python
import random #Loads Python’s random module into memory
```

```python
products = ["A","B","C","D","E"] #Creates a Python list in memory
lines = ["order_id,product,quantity,price"] #Creates a list with ONE string
                                            #That string represents the CSV header
                                            #One string = one CSV row

for i in range(1,1000): #range(1,1001) creates an iterable (not a list)
                        #Loop runs exactly 1000 times. Each iteration represents one incoming record
  order_id = i #Assigns integer i to order_id

  # Introduce missing product sometimes
  product = random.choice(products + [""])

  # Introduce bad quantity sometimes
  quantity = random.choice([1,2,3,4,5,"","abc"])

  # Introduce bad quantity sometimes
  price = random.choice([10,20,30,40,50,"xyz"])

  lines.append(f"{order_id},{product},{quantity},{price}")

csv_content = "\n\t".join(lines)

with open("sales_data_1000.csv","w") as file:
  file.write(csv_content)

print("sales_data_1000.csv created with 1000 rows")
```

```python
with open("sales_data_1000.csv","r") as file:
  for i,line in enumerate(file):
    if i < 10:
      print(line.strip())
```

```python
parsed_record = []
with open("sales_data_1000.csv","r") as file:
  header = file.readline().strip().split(",")

  for line_number, line in enumerate(file, start=2):
    line = line.strip()
    value = line.split(",")
    record = dict(zip(header, value))'''zip(header, value) pairs:("Date", "2024-01-01"),("Product", "Pen"),("Amount", "10")
                                        dict(...) converts pairs into a dictionary
                                        Final result:{"Date": "2024-01-01","Product": "Pen","Amount": "10"}'''

    record["_line_number"] = line_number '''Adds an extra key-value pair.Stores the original line number from the CSV file
                                            Useful for:
                                              Error tracking
                                              Debugging
                                              Data validation'''

    parsed_record.append(record)'''Adds the dictionary to the list. The list becomes a collection of structured records'''

print("Total parsed records:",len(parsed_record))
print("Sample Records:")
for r in parsed_record[:3]:
  print(r)
```

```python
clean_records = [] #Will store valid records
rejected_records= [] #Will store invalid records with error details

for record in parsed_record: #Loops through each dictionary in parsed_record
                             #Each record represents one CSV row
  errors = [] #Creates an empty list to store validation errors
              #Reset for each record

  try:
    record["order_id"] = int(record["order_id"]) '''Attempts to convert record["order_id"] to int
                                                  If conversion succeeds:
                                                    Value is replaced with an integer
                                                  If conversion fails:
                                                    Any exception is caught
                                                  Error message is appended
                                                    except: catches all exceptions (not best practice, but functional)'''
  except:
    errors.append("Order ID must be in integer")

    if not record["product"]:
      errors.append("Product is missing")

  try:
    record["quantity"] = int(record["quantity"])
  except:
    errors.append("Quantity must be in integer")

  try:
    record["price"] = float(record["price"])
  except:
    errors.append("Price must be in integer")

  if errors:
    rejected_records.append({
        "line":record["_line_number"],
        "record":record,
        "errors":errors
        })
  else:
    clean_records.append(record)
```

```python
with open("cleaned_sales_data.csv","w") as file:

  # Write header
  file.write("order_id, product, quantity, price\n")

  # Write each clean record
  for record in clean_records:
    line = f"{record['order_id']},{record['price']},{record['product']},{record['quantity']}\n"
    file.write(line)
print("Cleaned_sales_data.csv written successfully")
```

```python
with open("rejected_sales_data.csv","w") as file:

  file.write("line_number, errors, raw_records")

  for item in rejected_records:
    line_number = item["line"]
    errors = "|".join(item["errors"])
    raw_records = item["record"]

    file.write(f"{line_number}, {errors}, {raw_records}\n")

print("Rejected_Sales_data.csv written successfully")
```
