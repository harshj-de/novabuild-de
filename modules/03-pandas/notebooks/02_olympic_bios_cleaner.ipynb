Removing Dot '•' from the Used Names

```python
df["Used name"] = (
    df["Used name"]
      .astype(str) #Converts every value in the column to a Python string.
      .str.replace(r"[•·‧]", " ", regex=True) #Replaces any occurrence of these characters:[•·‧] " " replaces each matched character with a single space
      .str.strip()#Removes leading and trailing whitespace from each string.
)
```

```python
df.head()
```

Adding Spaces Between Names and string them in new column "display_name"

```python
df["display_name"] = (
    df["Used name"]
    .astype(str)
    .str.replace(r"[•·‧]", "", regex=True)
    .str.replace(r"([A-Z])([a-z])",r"\1 \2", regex=True)
    .str.strip()
)
```

Cleaning Measurements

Part	Meaning
(\d+)	one or more digits → height
\s*	optional spaces
cm	literal unit
/	separator
(\d+)	one or more digits → weight
kg	literal unit

```python
df[["height_cm","weight_kg"]]=(
    df["Measurements"].str.extract(r"(\d+)\s*cm\s*/\s*(\d+)\s*kg")
)
```

```python
df[["Measurements","height_cm","weight_kg"]].head(10)
```

Converting height_cm and weight_kg to numeric

```python
df['height_cm'] = pd.to_numeric(df['height_cm'], errors="coerce")
df['weight_kg'] = pd.to_numeric(df['weight_kg'], errors="coerce")
```

```python
df[['height_cm', 'weight_kg']].dtypes
```

Validate height_cm and weight_kg ranges

```python
df.loc[~df['height_cm'].between(120, 250), "height_cm"] = pd.NA
df.loc[~df['weight_kg'].between(30, 250), "weight_kg"] = pd.NA
```

```python
df[["height_cm", "weight_kg"]].isna().sum()
```

```python
df[["height_cm", "weight_kg"]].describe()
```

Cleaning the Born column, Extracting the Date String

```python
#June 30, 1985
#21 August 1986
#1997-03-14
born_extracted = df['Born'].str.extract(
    r"([A-Za-z]+\s+\d{1,2},\s+\d{4}|\d{1,2}+\s+[A-Za-z]+\s+\d{4}|\d{4}-\d{2}-\d{2})")
```

Convert extracted birth date text to datetime

```python
df["born_date"] = pd.to_datetime(born_extracted[0], errors="coerce")
```

```python
df[["Born", "born_date"]].head(10)
```

```python
df["born_date"].dtype
```

Extract birth location (city / region / country)

```python
location_extracted = df["Born"].str.extract(
    r"in\s+([^,]+),\s*([^()]+)\s*\(([^)]+)\)"
)

#1)in\s+
#Match the word in followed by space
#1)([^,]+)
#→ City
#Everything up to first comma
#3),\s*
#Comma + optional space
#4)([^()]+)
#→ Region
#Everything up to (
#5)\(([^)]+)\)
#→ Country
#Text inside parentheses
```

Assign to meaningful column names

```python
df[["birth_city", "birth_region", "birth_country"]] = location_extracted
```

Then clean whitespace:

```python
for col in ['birth_city','birth_region','birth_country']:
  df[col] = df[col].str.strip()
```

```python
df[["Born", "birth_city", "birth_region", "birth_country"]].head(10)
```

Born column drop

```python
df.drop(columns=["Born"], inplace=True)
```

```python
df.columns
```

Cleaning Affiliations

```python
aff_extracted = df["Affiliations"].str.extract(
    r"^([^,]+),\s*([^()]+)\s*\(([^)]+)\)$"
)
```

Assign extracted values

```python

df[["affiliation_name", "affiliation_city", "affiliation_country"]] = aff_extracted
```

Clean whitespace (only this)


```python
for col in ["affiliation_name", "affiliation_city", "affiliation_country"]:
    df[col] = df[col].str.strip()
```

```python
df[["Affiliations", "affiliation_name", "affiliation_city", "affiliation_country"]].head(10)
```

Country Normalization — FRA → France (and others)

```python
pd.concat([
    df["birth_country"],
    df["death_country"],
    df["affiliation_country"]
]).dropna().unique()
```

```python
country_map = {
    "FRA": "France",
    "GBR": "United Kingdom",
    "USA": "United States",
    "GER": "Germany",
    "ITA": "Italy",
    "ESP": "Spain"
}
```

```python
for col in ["birth_country", "death_country", "affiliation_country"]:
    df[col] = df[col].map(country_map)
```
